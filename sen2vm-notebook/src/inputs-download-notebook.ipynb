# GEOID folder copy from github

```python
# === GEOID ===
# This step is optional if the GEOID are already copied, or if you want to you use your own GEOID,
# It will copy the folder GEOID from "*/sen2vm-core/src/test/resources/DEM_GEOID" to PATH_TO_GEOID

import os
import sys
import shutil

if not os.path.exists(PATH_TO_GEOID):
    print("The folder where you want to download the GEOID is missing")
    sys.exit()   

GEOID_DIR = os.path.join(PATH_TO_GEOID, "GEOID")

if os.path.exists(GEOID_DIR):
    shutil.rmtree(GEOID_DIR)

os.makedirs(GEOID_DIR)

# Notebook location (NOT relative to CWD)
notebook_dir = os.getcwd()

# Relative path to DEM_GEOID from notebook
internal_geoid_dir = os.path.abspath(os.path.join(
    notebook_dir,
    "..", "..", "src", "test", "resources", "DEM_GEOID"
))

for f in os.listdir(internal_geoid_dir):
    src = os.path.join(internal_geoid_dir, f)
    dst = os.path.join(GEOID_DIR, f)
    shutil.copy(src, dst)
print("Done copying")
```
